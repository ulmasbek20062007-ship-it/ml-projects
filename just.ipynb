```python
import pandas as pd

data = [
    [1, 'Nina', 45, 2],
    [2, 'Omar', 38, None],
    [3, 'Priya', 52, 2],
    [4, 'Quinn', 30, 1],
]
players = pd.DataFrame(data, columns=['id', 'name', 'age', 'coachId'])
print(players)
```

Task: find every player who is older than their coach. Return just the player's name, renamed to Player

```python
merged  = pd.merge(players, players, left_on = 'coachId', right_on = 'id', suffixes = ('', '_coach'))
result = merged[merged['age'] > merged['age_coach']]
result = result[['name']].rename(columns = {'name':'Player'})
print(result)
```

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def maxDepth(root):
    if not root:
        return 0
    else:
        return 1 + max(maxDepth(root.left), maxDepth(root.right))
    
# build a small tree:
#       1
#      / \
#     2   3
#    /
#   4
root = TreeNode(1, TreeNode(2, TreeNode(4)), TreeNode(3))

print(maxDepth(root))  # expected: 3
```

```python
from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def levelOrder(root):
    if not root:
        return
    
    queue = deque([root])
    res= []

    while queue:
        k = len(queue)
        gr = []
        for _ in range(k):
            node = queue.popleft()
            gr.append(node.val)

            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        res.append(gr)
    return res

# build the tree:
#       1
#      / \
#     2   3
#    /
#   4
root = TreeNode(1, TreeNode(2, TreeNode(4)), TreeNode(3))

print(levelOrder(root))  # expected: [[1], [2, 3], [4]]
```

```python
def isvalid(grid, row, col):
    r = len(grid)
    c = len(grid[0])
    if 0 <= row < r and 0 <= c < col:
        return True
    else:
        return False
```

```python
def numIslands(grid):
    rows = len(grid)
    cols = len(grid[0])
    visited = set()
    count = 0

    def dfs(row, col):
        if (row < 0 or col < 0 or grid[row][col] == '0' or row >= rows or col >= cols or (row, col) in visited):
            return
        visited.add((row, col))

        dfs(row-1, col)
        dfs(row, col-1)
        dfs(row, col+1)
        dfs(row+1, col)

    for row in rows:
        for col in cols:
            if grid[row][col] == '1' and (row, col) not in visited:
                count += 1
                dfs(row, col)

    return count




```
